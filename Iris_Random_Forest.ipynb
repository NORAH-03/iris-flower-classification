# Step 1: Import Required Libraries

```python
# ==========================================
# Step 1: Import Required Libraries
# ==========================================

# Numerical computing
import numpy as np

# Data manipulation
import pandas as pd

# Data visualization
import matplotlib.pyplot as plt

# Load Iris dataset
from sklearn.datasets import load_iris

# Split dataset
from sklearn.model_selection import train_test_split

# Random Forest model
from sklearn.ensemble import RandomForestClassifier

# Evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

# Step 2: Load the Iris Dataset

```python
# ==========================================
# Step 2: Load the Iris Dataset
# ==========================================

# Load the built-in Iris dataset
iris = load_iris()

# Separate the features (X) and target labels (y)
X = iris.data
y = iris.target
```

# Step 3: Explore the Dataset

```python
# ==========================================
# Step 3: Dataset Overview
# ==========================================

print("Dataset Overview")
print("-" * 40)

print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
print("Dataset shape:", X.shape)

print("\nFeature Names:")
print(iris.feature_names)

print("\nTarget Names:")
print(iris.target_names)
```

```python
# ==========================================
# Create a Pandas DataFrame
# ==========================================

# Create a DataFrame from the feature data
df = pd.DataFrame(X, columns=iris.feature_names)

# Add the target labels
df["Species"] = y

print("First Five Rows of the Dataset")
print("-" * 40)

df.head()
```

```python
# ==========================================
# Statistical Summary
# ==========================================

print("Statistical Summary")
print("-" * 40)

df.describe()
```

# Dataset Explanation

### What is the problem?

The Iris dataset is used to classify iris flowers into one of three different species based on their physical measurements.

### Input Features

The model uses four numerical features:

- Sepal Length (cm)
- Sepal Width (cm)
- Petal Length (cm)
- Petal Width (cm)

### Target Variable

The target variable is the species of the flower:

- Setosa
- Versicolor
- Virginica

### Why is this a Classification Problem?

The objective is to predict one of three predefined categories (flower species). Since the output is categorical rather than continuous, this is a multiclass classification problem.

# Step 4: Split the Dataset into Training and Testing Sets

```python
# ==========================================
# Step 4: Split the Dataset
# ==========================================

# Split the dataset into training (80%) and testing (20%) sets.
# random_state=42 ensures reproducible results.
# stratify=y maintains the class distribution in both sets.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set Shape:", X_train.shape)
print("Testing Set Shape:", X_test.shape)
```

## Why do we split the dataset?

Machine learning models should be evaluated on data they have never seen before.

- The **training set** is used to teach the model.
- The **testing set** is used to evaluate how well the model generalizes to new data.

Using separate datasets helps detect overfitting and provides a fair estimate of the model's performance.

# Step 5: Build the Random Forest Classifier

```python
# ==========================================
# Step 5: Build the Random Forest Model
# ==========================================

# Create a Random Forest classifier.
# n_estimators specifies the number of decision trees.
# random_state ensures reproducible results.

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

print(rf_model)
```

## Random Forest Parameters

### n_estimators

This parameter specifies the number of decision trees in the forest. More trees generally improve performance but increase training time.

### random_state

This parameter ensures that the results are reproducible. Using the same random state produces the same train/test split and model each time the notebook is run.


# Step 6: Train the Model

```python
# ==========================================
# Step 6: Train the Model
# ==========================================

# Train the Random Forest model using the training data.

rf_model.fit(X_train, y_train)

print("The Random Forest model has been trained successfully.")
```

# Step 7: Make Predictions

```python
# ==========================================
# Step 7: Make Predictions
# ==========================================

# Predict the labels for both the training and testing datasets.

train_predictions = rf_model.predict(X_train)
test_predictions = rf_model.predict(X_test)

print("Training predictions completed.")
print("Testing predictions completed.")
```

```python
# ==========================================
# Compare Actual vs Predicted Labels
# ==========================================

comparison = pd.DataFrame({
    "Actual": y_test,
    "Predicted": test_predictions
})

comparison.head(10)
```

# Step 8: Evaluate the Model

```python
# ==========================================
# Step 8: Evaluate the Model
# ==========================================

# Calculate the accuracy on both the training and testing datasets.

train_accuracy = accuracy_score(y_train, train_predictions)
test_accuracy = accuracy_score(y_test, test_predictions)

print("Model Accuracy")
print("-" * 40)

print(f"Training Accuracy: {train_accuracy:.2%}")
print(f"Testing Accuracy: {test_accuracy:.2%}")
```

```python
# ==========================================
# Classification Report
# ==========================================

print("Classification Report")
print("-" * 40)

print(classification_report(
    y_test,
    test_predictions,
    target_names=iris.target_names
))
```

```python
# ==========================================
# Confusion Matrix
# ==========================================

cm = confusion_matrix(y_test, test_predictions)

print("Confusion Matrix")
print("-" * 40)

print(cm)
```

# Step 9: Visualize the Confusion Matrix

```python
# ==========================================
# Plot Confusion Matrix
# ==========================================

plt.figure(figsize=(6, 5))

plt.imshow(cm, cmap="Blues")

plt.title("Confusion Matrix")

plt.colorbar()

plt.xticks(
    range(len(iris.target_names)),
    iris.target_names,
    rotation=45
)

plt.yticks(
    range(len(iris.target_names)),
    iris.target_names
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")

# Display values inside each cell
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center",
            color="black"
        )

plt.tight_layout()

plt.show()
```

# Step 10: Feature Importance

```python
# ==========================================
# Feature Importance
# ==========================================

feature_importance = pd.DataFrame({
    "Feature": iris.feature_names,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

feature_importance
```

```python
# ==========================================
# Plot Feature Importance
# ==========================================

plt.figure(figsize=(8, 5))

plt.bar(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.title("Feature Importance")

plt.xlabel("Features")
plt.ylabel("Importance Score")

plt.xticks(rotation=20)

plt.tight_layout()

plt.show()
```

# Step 11: Compare Training and Testing Accuracy

```python
# ==========================================
# Compare Model Performance
# ==========================================

print(f"Training Accuracy: {train_accuracy:.2%}")
print(f"Testing Accuracy: {test_accuracy:.2%}")

difference = train_accuracy - test_accuracy

print(f"\nAccuracy Difference: {difference:.2%}")
```

```python
# ==========================================
# Determine Model Fit
# ==========================================

if train_accuracy > test_accuracy + 0.10:
    print("Model Status: Overfitting")
    print("Reason: Training accuracy is significantly higher than testing accuracy.")

elif train_accuracy < 0.70 and test_accuracy < 0.70:
    print("Model Status: Underfitting")
    print("Reason: Both training and testing accuracies are low.")

else:
    print("Model Status: Well-fitted")
    print("Reason: Training and testing accuracies are similar, indicating good generalization.")
```

## Model Interpretation

The Random Forest classifier achieved high accuracy on both the training and testing datasets. The small difference between the two accuracy scores indicates that the model generalizes well to unseen data. Therefore, the model is considered **well-fitted** rather than overfitted or underfitted.

# Step 12: Parameter Experiment (Effect of n_estimators)

```python
# ==========================================
# Step 12: Parameter Experiment
# ==========================================

# Test different values of n_estimators
estimators = [10, 50, 100, 200]

# Store the testing accuracy for each model
accuracy_results = []

for n in estimators:
    model = RandomForestClassifier(
        n_estimators=n,
        random_state=42
    )

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    accuracy = accuracy_score(y_test, predictions)

    accuracy_results.append(accuracy)
```

```python
# ==========================================
# Display Results
# ==========================================

results_df = pd.DataFrame({
    "Number of Trees (n_estimators)": estimators,
    "Testing Accuracy": accuracy_results
})

results_df
```

# Step 13: Plot the Parameter Experiment

```python
# ==========================================
# Plot Accuracy vs Number of Trees
# ==========================================

plt.figure(figsize=(8,5))

plt.plot(
    estimators,
    accuracy_results,
    marker="o",
    linewidth=2
)

plt.title("Effect of n_estimators on Model Accuracy")

plt.xlabel("Number of Trees (n_estimators)")

plt.ylabel("Testing Accuracy")

plt.xticks(estimators)

plt.grid(True)

plt.show()
```

## Discussion of Parameter Experiment

The parameter experiment evaluated the effect of changing the number of trees (`n_estimators`) in the Random Forest model.

As the number of trees increased, the model generally became more stable and achieved high prediction accuracy. Increasing the number of trees beyond a certain point did not significantly improve the accuracy, indicating that the model had already learned the dataset effectively.

For this dataset, **100 trees** provided excellent performance while maintaining reasonable computational cost, making it a suitable choice for this project.
